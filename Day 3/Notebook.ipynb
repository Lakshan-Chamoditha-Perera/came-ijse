# Python Dictionary (`dict`) Notes

A dictionary stores data as **key-value pairs**.

```python
data = {
    "name": "Sugar",
    "weight": "1kg",
    "price": 135.50,
    "price": 135.50
}
```

## 1. Duplicate Keys Are Not Allowed

```python
print(data)
```

### Output

```python
{'name': 'Sugar', 'weight': '1kg', 'price': 135.5}
```

### Notes

* Dictionaries cannot have duplicate keys.
* If the same key appears multiple times, Python keeps only the **last occurrence**.
* In this example, only one `"price"` key exists in the final dictionary.

---

## 2. Updating a Value

```python
data["name"] = "Rice"
print(data)
```

### Output

```python
{'name': 'Rice', 'weight': '1kg', 'price': 135.5}
```

### Notes

* Assigning a value to an existing key updates that key's value.
* `"name"` changed from `"Sugar"` to `"Rice"`.

---

## 3. Adding a New Key-Value Pair

```python
data["desc"] = "sample"
print(data)
```

### Output

```python
{
    'name': 'Rice',
    'weight': '1kg',
    'price': 135.5,
    'desc': 'sample'
}
```

### Notes

* If the key does not exist, Python adds it to the dictionary.
* New items are added at the end of the dictionary (Python 3.7+ preserves insertion order).

---

## Summary

| Operation         | Code                      | Result                     |
| ----------------- | ------------------------- | -------------------------- |
| Create dictionary | `data = {...}`            | Creates a dictionary       |
| Access value      | `data["name"]`            | Gets the value of `"name"` |
| Update value      | `data["name"] = "Rice"`   | Changes an existing value  |
| Add new item      | `data["desc"] = "sample"` | Adds a new key-value pair  |
| Duplicate key     | `"price"` repeated        | Last value is kept         |

## Final Dictionary

```python
{
    'name': 'Rice',
    'weight': '1kg',
    'price': 135.5,
    'desc': 'sample'
}
```

> **Key Point:** Dictionary keys must be unique. Existing keys are updated, while new keys are added.


````python
data = {
    "name":"Sugar",
    "weight":"1kg",
    "price":135.50
}

print(data)
print(type(data))
````

# Loops 


## For Loop

A **for loop** is used to repeat a block of code for each item in a sequence. It allows you to perform the same action multiple times without writing the same code repeatedly.

For loops are commonly used when working with collections of data such as lists, strings, tuples, dictionaries, and ranges. During each iteration, the loop processes one item at a time until all items have been handled.

### Why Use a For Loop?

* Reduces repetitive code.
* Makes programs shorter and easier to read.
* Helps process multiple values efficiently.
* Useful for traversing collections and performing repeated tasks.

### Key Concepts

* **Iteration**: One execution of the loop body.
* **Sequence**: A collection of items that can be traversed one by one.
* **Loop Variable**: Holds the current item being processed during each iteration.

### Common Uses

* Displaying items in a collection.
* Calculating totals and averages.
* Searching for specific values.
* Processing text character by character.
* Repeating tasks a fixed number of times.

A for loop continues until every item in the sequence has been processed, making it one of the most commonly used control structures in Python.


````python
animals = ["dog", "cat", "rabbit"]
for animal in animals:
    print(f"{animal}")
````

# Range Function 

The range() function generates a sequence of numbers. It is commonly used with for loops to control the number of iterations.

````python
# Range function
for i in range(0,10,2):   #range(start_value, stop_value, step) 
    print(i)
````

## Range with Start, Stop, and Step
```python
for i in range(0, 10, 2):   # range(start_value, stop_value, step)
    print(i)
```

Explanation
0 → Starting value (included).
10 → Stopping value (not included).
2 → Step value (increment by 2).

````python
for i in range(0, 10, 2):   # range(start_value, stop_value, step)
    print(i)
````

## Range with a Single Value
```python
for item in range(5):
    print(item)
```

Explanation :
When a single value is provided, it is treated as the stopping value.
- The sequence starts from 0 by default.
- The stop value is not included.

````python
for item in range(5): 
    print(item)
````

## Range with Start and Stop Values
```python
for item in range(1, 10):
    print(item)
```
Explanation
<br>
- 1 → Starting value (included).
- 10 → Stopping value (not included).
- The default step value is 1.

````python
# Passing two values to range function
for item in range(1, 10):
    print(item)
````

# Enumerate Function

The ```enumerate()``` function is used when you need both the index and the value while looping through a sequence.

````python
labals = ["a", "b", "c"] 

x = enumerate(labals)

print(x) # prints the object
print(list(x)) # prints index and labels 
````

````python
for index,label in enumerate(labals):
    print(f"index: {index}, label: {label}")
````

````python
scores = [0.85, 0.91, 0.78, None, 0.92, 0.2, 0.88]
for score in scores:
    if type(score) is float:
        print(score)
````

````python
## continue statement
for score in scores:
    if score is None:
        print("Score is None, skipping...")
        continue
    print(f"Score: {score}")
````

````python
## break statement
for score in scores:
    
    if score is None:
        print("Score is None, skipping...")
        continue

    if score < 0.5:
        print(f"Low score: {score} breaking the loop.")
        break
    
    print(f"Score: {score}")
````
