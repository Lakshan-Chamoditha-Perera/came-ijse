```python
# %pip install pandas matplotlib seaborn
import pandas as pd
import matplotlib as plt
import seaborn as sns
```

```python
df = pd.DataFrame(
    {
        "Name": [
            "Alice",
            "Bob",
            "Charlie",
            "David",
            "Eve",
            "Frank",
            "Grace",
            "Hannah",
            "Ian",
            "Jack",
        ],
        "Age": [22, 25, 30, 35, 40, 45, 50, 55, 60, 65],
        "Score": [0.85, 0.90, 0.95, 0.80, 0.75, 0.70, 0.65, 0.60, 0.55, 0.50],
        "City": [
            "New York",
            "Los Angeles",
            "Chicago",
            "New York",
            "Phoenix",
            "Philadelphia",
            "Chicago",
            "San Diego",
            "Chicago",
            "Philadelphia",
        ],
    }
)
```

```python
#histogram of Age using matplotlib

plt.pyplot.hist(df["Age"], bins=10, color="blue", alpha=0.7)
plt.pyplot.title("Histogram of Age")
plt.pyplot.xlabel("Age")
plt.pyplot.ylabel("Frequency")
plt.pyplot.show()
```

```python
city_mean = df.groupby("City")["Score"].mean()
print(city_mean)
```

```python
city_mean.plot(kind="bar", color="green", alpha=0.7)
plt.pyplot.title("Average Score by City")
plt.pyplot.xlabel("City")
plt.pyplot.ylabel("Average Score")
plt.pyplot.show()
```

## Scatter plot
A scatter plot (or scatter diagram) is a type of graph that displays the relationship between two continuous (numerical) variables using Cartesian coordinates. Each point represents an observation with values on both axes.

A scatter plot is an exploratory data analysis (EDA) tool that helps you understand relationships between numerical features. While a strong positive or negative linear relationship may indicate that two features contain similar information, you should not automatically drop one. Instead, 

 consider:

- How strong the correlation is.
- Whether the features represent different real-world concepts.
- Which machine learning algorithm you're using.
- Whether removing a feature improves model performance and interpretability.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
plt.scatter(df["Age"], df["Score"])

plt.title("Age vs Score")
plt.xlabel("Age")
plt.ylabel("Score")

plt.grid(True)
plt.show()
```

## Box Plot
A box plot (box-and-whisker plot) is a visualization technique used to analyze continuous numerical data through summary statistics.

It provides a compact summary of the distribution of a numerical feature using five important statistical values:
- Minimum value
- First Quartile (Q1 - 25th percentile)
- Median (Q2 - 50th percentile)
- Third Quartile (Q3 - 75th percentile)

```python
# Box plot of age vs city using seaborn
plt.figure()
sns.boxplot(x="City", y="Age", data=df)
plt.title("Box Plot of Age by City")
plt.xlabel("City")
plt.ylabel("Age")
plt.xticks(rotation=45)
plt.show()
```
