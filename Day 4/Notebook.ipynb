# Functions in Python

## Definition of a Function

A **function** is a reusable block of code that performs a specific task. Functions help make programs more organized, readable, and easier to maintain.

### Syntax

```python
def function_name(parameters):
    # function body
    return value
```

### Example

```python
def greet(name):
    return f"Hello, {name}!"

print(greet("Alice"))
```

**Output:**

```text
Hello, Alice!
```

## Functions Without a Class

In Python, functions can be defined and used independently without creating a class. This is different from some other programming languages where methods are usually defined inside classes.

### Example

```python
def add(a, b):
    return a + b

result = add(5, 3)
print(result)
```

**Output:**

```text
8
```

### Key Points

* Functions are defined using the `def` keyword.
* Functions can accept parameters.
* Functions can return values using the `return` statement.
* Python allows standalone functions without requiring a class.


````python
def calculate_accuracy(correct_labels, total_labels):
    accuracy = (correct_labels / total_labels)
    return accuracy

accuracy_value = calculate_accuracy(85, 100)
print(f"Accuracy: {accuracy_value * 100:.2f}%")
````

## Default and Keyword Arguments

### 1. Default Arguments

Default arguments allow you to assign a default value to a parameter in a function definition. If the caller does not provide a value for that parameter, Python uses the default value.

#### How It Works

You assign a value to a parameter using the `=` operator in the function definition.

#### Example

```python
def greet(name, message="Hello"):
    print(f"{message}, {name}!")
```

Here, `message` has a default value of `"Hello"`.

```python
greet("Alice")
```

Output:

```python
Hello, Alice!
```

Since no value was provided for `message`, Python used the default value.

#### Function Definition Rule

Parameters without default values must come before parameters with default values.

**Correct:**

```python
def create_user(username, email, is_active=True):
    pass
```

**Incorrect:**

```python
def create_user(is_active=True, username, email):
    pass
```

Python requires all default parameters to appear after non-default parameters.


````python
def calculate_accuracy(correct_labels, total_labels=10):
    accuracy = (correct_labels / total_labels)
    return accuracy

accuracy_value = calculate_accuracy(1)
print(f"Accuracy: {accuracy_value * 100:.2f}%")
````

### 2. Keyword Arguments

Keyword arguments allow you to pass values to a function by explicitly specifying the parameter names.

#### How It Works

You assign a value to a parameter using the `=` operator during the function call.

#### Example

```python
def greet(name, message="Hello"):
    print(f"{message}, {name}!")
```

Using keyword arguments:

```python
greet("Alice", message="Hi")
greet(name="Alice", message="Hi")
```

Output:

```python
Hi, Alice!
Hi, Alice!
```

#### Rules for Keyword Arguments

* Keyword arguments can be passed in any order because Python matches them by parameter name.
* Positional arguments must come before keyword arguments.

**Correct:**

```python
greet("Alice", message="Hi")
```

**Incorrect:**

```python
greet(name="Alice", "Hi")
```

This causes a syntax error because a positional argument appears after a keyword argument.

#### Benefit of Keyword Arguments

They make function calls more readable and allow you to specify only the arguments you want to change, especially when a function has many parameters.

```python
create_user(
    username="alice",
    email="alice@example.com",
    is_active=False
)
```

The parameter names clearly indicate what each value represents.

````python
def calculate_area(length, width):
    area = length * width
    return area

# Using keyword arguments
room_area = calculate_area(width=5, length=8)

print(f"Area: {room_area} square meters")
````

## Arbitrary Arguments

### Arbitrary Positional Arguments (`*args`)

Sometimes you don’t know in advance how many arguments will be passed into a function. Python allows you to handle this using `*args`.

### How it works

* The `*` before a parameter name allows the function to accept **any number of positional arguments**.
* All extra arguments are collected into a **tuple**.

#### Example 1: Basic `*args`

```python id="args1"
def add_numbers(*args):
    total = 0
    for num in args:
        total += num
    return total

result = add_numbers(2, 4, 6, 8)

print(f"Sum: {result}")
```

**Output:**

```
Sum: 20
```

---

#### Example 2: Flexible inputs

```python id="args2"
def show_students(*args):
    print("Students list:")
    for student in args:
        print(student)

show_students("Alice", "Bob", "Charlie")
```

**Output:**

```
Students list:
Alice
Bob
Charlie
```

---

#### Key Points

* `*args` collects multiple positional arguments into a tuple
* You can pass **any number of values**
* It must come after normal positional parameters (if any)

#### Example with normal parameter + `*args`

```python id="args3"
def greet(greeting, *names):
    for name in names:
        print(f"{greeting}, {name}!")

greet("Hello", "Alice", "Bob", "Charlie")
```

**Output:**

```
Hello, Alice!
Hello, Bob!
Hello, Charlie!
```

---

#### Summary

`*args` makes functions flexible by allowing them to accept an unlimited number of positional arguments, which are handled as a tuple inside the function.


````python
def calculate_average(*args):
    try:
        if not args:
            raise ValueError("At least one number must be provided")

        for number in args:
            if not isinstance(number, (int, float)):
                raise ValueError("All inputs must be numbers")

        total = sum(args)
        count = len(args)
        average = total / count
        return average
    except ValueError as e:
        print(f"Error: {e}")


average_value = calculate_average(1, 2, 3)
print(f"Average: {average_value}\n")

average_value = calculate_average()
print(f"Average: {average_value}\n")

average_value = calculate_average("", 0)  
print(f"Average: {average_value}\n")
````

### Arbitrary Keyword Arguments

**Arbitrary Keyword Arguments** in Python let you pass a variable number of named arguments to a function using `**kwargs`.

#### Basic idea

* `*args` → collects extra positional arguments (tuple)
* `**kwargs` → collects extra keyword arguments (dictionary)

#### Example

```python
def show_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

show_info(name="Alice", age=25, city="Colombo")
```

#### Output

```
name: Alice
age: 25
city: Colombo
```

#### How it works

Inside the function:

* `kwargs` becomes a **dictionary**
* Keys are argument names
* Values are argument values

So in the example:

```python
kwargs = {
    "name": "Alice",
    "age": 25,
    "city": "Colombo"
}
```

#### Using with other parameters

```python
def greet(prefix, **kwargs):
    print(prefix)
    for k, v in kwargs.items():
        print(k, v)

greet("User Data:", name="Bob", role="Admin")
```

#### Key points

* Must come **after positional parameters**
* Can be combined with `*args`
* Useful for flexible APIs and configuration-style functions




````python
def calculate_average(**kwargs):
    try:
        if not kwargs:
            raise ValueError("At least one number must be provided")

        for key, value in kwargs.items():
            print(f"Key: {key}, Value: {value}")
            if not isinstance(value, (int, float)):
                raise ValueError(f"Value for '{key}' must be a number")

        average = sum(kwargs.values()) / len(kwargs)
        return average

    except ValueError as e:
        print(f"Error: {e}")
        return None


average_value = calculate_average(a=1, b=2, c=3)
print(f"Average: {average_value}\n")

average_value = calculate_average()
print(f"Average: {average_value}\n")

average_value = calculate_average(x="", y=0)
print(f"Average: {average_value}\n")
````

# DocStrings

Docstrings are used in **functions, classes, modules, and methods** to describe their purpose, behavior, parameters, and return values. They help developers understand what the code does without reading the full implementation.

They are **not like comments**. Comments are ignored by the interpreter at runtime, but docstrings are not ignored—they are stored in memory and attached to the object.

Docstrings can be accessed using tools like:

* `help(object)`
* `object.__doc__`

### Key Features:

* Written as a string literal (usually triple quotes `""" ... """`)
* Placed immediately after the definition of a function, class, or module
* Used for documentation and introspection
* Can be used by documentation generators (like Sphinx)

### Example Topics Covered in Docstrings:

* Function purpose
* Parameters (inputs)
* Return values (outputs)
* Class behavior and attributes
* Module overview

### Example (Google Style):

```python
def add(a, b):
    """
    Returns the sum of two numbers.

    Parameters:
        a (int or float): First number
        b (int or float): Second number

    Returns:
        int or float: Sum of a and b
    """
    return a + b
```

Docstrings are a key part of writing clean, readable, and maintainable Python code.


````python
def find_maximum(*args):
    """
    Find the maximum value from a set of numeric inputs.

    This function accepts multiple numeric values and returns the largest one.
    It validates that:
    - At least one value is provided
    - All values are int or float

    Parameters:
        *args: Variable number of numeric values (int or float)

    Returns:
        int or float: The maximum value among inputs
        None: If input is invalid or empty
    """
    try:
        if not args:
            raise ValueError("At least one number must be provided")

        for i, value in enumerate(args):
            print(f"Index: {i}, Value: {value}")
            if not isinstance(value, (int, float)):
                raise ValueError(f"Value at position {i} must be a number")

        return max(args)

    except ValueError as e:
        print(f"Error: {e}")
        return None

print(find_maximum(10, 5, 8, 20))
````

````python
print(f"Docstring for find_maximum:\n{find_maximum.__doc__}")
````

````python
print(f"Docstring for find_maximum:\n{help(find_maximum)}")
````
