# Functions in Python

## Definition of a Function

A **function** is a reusable block of code that performs a specific task. Functions help make programs more organized, readable, and easier to maintain.

### Syntax

```python
def function_name(parameters):
    # function body
    return value
```

### Example

```python
def greet(name):
    return f"Hello, {name}!"

print(greet("Alice"))
```

**Output:**

```text
Hello, Alice!
```

## Functions Without a Class

In Python, functions can be defined and used independently without creating a class. This is different from some other programming languages where methods are usually defined inside classes.

### Example

```python
def add(a, b):
    return a + b

result = add(5, 3)
print(result)
```

**Output:**

```text
8
```

### Key Points

* Functions are defined using the `def` keyword.
* Functions can accept parameters.
* Functions can return values using the `return` statement.
* Python allows standalone functions without requiring a class.


````python
def calculate_accuracy(correct_labels, total_labels):
    accuracy = (correct_labels / total_labels)
    return accuracy

accuracy_value = calculate_accuracy(85, 100)
print(f"Accuracy: {accuracy_value * 100:.2f}%")
````

## Default and Keyword Arguments

These are used during a function call and allow you to pass arguments by explicitly naming the parameter.

* **How it works:** You assign a value to a parameter using the `=` operator.
* **The rule:** Keyword arguments must come after any positional arguments in a function call.
* **The benefit:** The order of keyword arguments does not matter because Python matches each value to the parameter name.

**Example:**

```python
def greet(name, message="Hello"):
    print(f"{message}, {name}!")

greet("Alice")                    # Positional argument
greet("Alice", message="Hi")      # Positional + keyword argument
greet(name="Alice", message="Hi") # All keyword arguments
```

### Function Definition Guidelines

When defining a function:

* Place **non-default (required) parameters** first.
* Place **default (optional) parameters** after the required parameters.

**Correct:**

```python
def create_user(username, email, is_active=True):
    pass
```

**Incorrect:**

```python
def create_user(is_active=True, username, email):
    pass
```

Python requires all parameters with default values to appear **after** parameters without default values.


````python
def calculate_accuracy(correct_labels, total_labels=10):
    accuracy = (correct_labels / total_labels)
    return accuracy

accuracy_value = calculate_accuracy(1)
print(f"Accuracy: {accuracy_value * 100:.2f}%")
````
