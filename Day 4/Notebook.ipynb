# Functions in Python

## Definition of a Function

A **function** is a reusable block of code that performs a specific task. Functions help make programs more organized, readable, and easier to maintain.

### Syntax

```python
def function_name(parameters):
    # function body
    return value
```

### Example

```python
def greet(name):
    return f"Hello, {name}!"

print(greet("Alice"))
```

**Output:**

```text
Hello, Alice!
```

## Functions Without a Class

In Python, functions can be defined and used independently without creating a class. This is different from some other programming languages where methods are usually defined inside classes.

### Example

```python
def add(a, b):
    return a + b

result = add(5, 3)
print(result)
```

**Output:**

```text
8
```

### Key Points

* Functions are defined using the `def` keyword.
* Functions can accept parameters.
* Functions can return values using the `return` statement.
* Python allows standalone functions without requiring a class.


````python
def calculate_accuracy(correct_labels, total_labels):
    accuracy = (correct_labels / total_labels)
    return accuracy

accuracy_value = calculate_accuracy(85, 100)
print(f"Accuracy: {accuracy_value * 100:.2f}%")
````

## Default and Keyword Arguments

### 1. Default Arguments

Default arguments allow you to assign a default value to a parameter in a function definition. If the caller does not provide a value for that parameter, Python uses the default value.

#### How It Works

You assign a value to a parameter using the `=` operator in the function definition.

#### Example

```python
def greet(name, message="Hello"):
    print(f"{message}, {name}!")
```

Here, `message` has a default value of `"Hello"`.

```python
greet("Alice")
```

Output:

```python
Hello, Alice!
```

Since no value was provided for `message`, Python used the default value.

#### Function Definition Rule

Parameters without default values must come before parameters with default values.

**Correct:**

```python
def create_user(username, email, is_active=True):
    pass
```

**Incorrect:**

```python
def create_user(is_active=True, username, email):
    pass
```

Python requires all default parameters to appear after non-default parameters.


````python
def calculate_accuracy(correct_labels, total_labels=10):
    accuracy = (correct_labels / total_labels)
    return accuracy

accuracy_value = calculate_accuracy(1)
print(f"Accuracy: {accuracy_value * 100:.2f}%")
````


---


### 2. Keyword Arguments

Keyword arguments allow you to pass values to a function by explicitly specifying the parameter names.

#### How It Works

You assign a value to a parameter using the `=` operator during the function call.

#### Example

```python
def greet(name, message="Hello"):
    print(f"{message}, {name}!")
```

Using keyword arguments:

```python
greet("Alice", message="Hi")
greet(name="Alice", message="Hi")
```

Output:

```python
Hi, Alice!
Hi, Alice!
```

#### Rules for Keyword Arguments

* Keyword arguments can be passed in any order because Python matches them by parameter name.
* Positional arguments must come before keyword arguments.

**Correct:**

```python
greet("Alice", message="Hi")
```

**Incorrect:**

```python
greet(name="Alice", "Hi")
```

This causes a syntax error because a positional argument appears after a keyword argument.

#### Benefit of Keyword Arguments

They make function calls more readable and allow you to specify only the arguments you want to change, especially when a function has many parameters.

```python
create_user(
    username="alice",
    email="alice@example.com",
    is_active=False
)
```

The parameter names clearly indicate what each value represents.

````python
def calculate_area(length, width):
    area = length * width
    return area

# Using keyword arguments
room_area = calculate_area(width=5, length=8)

print(f"Area: {room_area} square meters")
````

## Arbitrary Arguments

### Arbitrary Positional Arguments (`*args`)

Sometimes you don’t know in advance how many arguments will be passed into a function. Python allows you to handle this using `*args`.

### How it works

* The `*` before a parameter name allows the function to accept **any number of positional arguments**.
* All extra arguments are collected into a **tuple**.

#### Example 1: Basic `*args`

```python id="args1"
def add_numbers(*args):
    total = 0
    for num in args:
        total += num
    return total

result = add_numbers(2, 4, 6, 8)

print(f"Sum: {result}")
```

**Output:**

```
Sum: 20
```

---

#### Example 2: Flexible inputs

```python id="args2"
def show_students(*args):
    print("Students list:")
    for student in args:
        print(student)

show_students("Alice", "Bob", "Charlie")
```

**Output:**

```
Students list:
Alice
Bob
Charlie
```

---

#### Key Points

* `*args` collects multiple positional arguments into a tuple
* You can pass **any number of values**
* It must come after normal positional parameters (if any)

#### Example with normal parameter + `*args`

```python id="args3"
def greet(greeting, *names):
    for name in names:
        print(f"{greeting}, {name}!")

greet("Hello", "Alice", "Bob", "Charlie")
```

**Output:**

```
Hello, Alice!
Hello, Bob!
Hello, Charlie!
```

---

#### Summary

`*args` makes functions flexible by allowing them to accept an unlimited number of positional arguments, which are handled as a tuple inside the function.


````python
def calculate_average(*args):
    try:
        if not args:
            raise ValueError("At least one number must be provided")

        for number in args:
            if not isinstance(number, (int, float)):
                raise ValueError("All inputs must be numbers")

        total = sum(args)
        count = len(args)
        average = total / count
        return average
    except ValueError as e:
        print(f"Error: {e}")


average_value = calculate_average(1, 2, 3)
print(f"Average: {average_value}\n")

average_value = calculate_average()
print(f"Average: {average_value}\n")

average_value = calculate_average("", 0)  
print(f"Average: {average_value}\n")
````

### Arbitrary Keyword Arguments

**Arbitrary Keyword Arguments** in Python let you pass a variable number of named arguments to a function using `**kwargs`.

#### Basic idea

* `*args` → collects extra positional arguments (tuple)
* `**kwargs` → collects extra keyword arguments (dictionary)

#### Example

```python
def show_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

show_info(name="Alice", age=25, city="Colombo")
```

#### Output

```
name: Alice
age: 25
city: Colombo
```

#### How it works

Inside the function:

* `kwargs` becomes a **dictionary**
* Keys are argument names
* Values are argument values

So in the example:

```python
kwargs = {
    "name": "Alice",
    "age": 25,
    "city": "Colombo"
}
```

#### Using with other parameters

```python
def greet(prefix, **kwargs):
    print(prefix)
    for k, v in kwargs.items():
        print(k, v)

greet("User Data:", name="Bob", role="Admin")
```

#### Key points

* Must come **after positional parameters**
* Can be combined with `*args`
* Useful for flexible APIs and configuration-style functions




````python
def calculate_average(**kwargs):
    try:
        if not kwargs:
            raise ValueError("At least one number must be provided")

        for key, value in kwargs.items():
            print(f"Key: {key}, Value: {value}")
            if not isinstance(value, (int, float)):
                raise ValueError(f"Value for '{key}' must be a number")

        average = sum(kwargs.values()) / len(kwargs)
        return average

    except ValueError as e:
        print(f"Error: {e}")
        return None


average_value = calculate_average(a=1, b=2, c=3)
print(f"Average: {average_value}\n")

average_value = calculate_average()
print(f"Average: {average_value}\n")

average_value = calculate_average(x="", y=0)
print(f"Average: {average_value}\n")
````

# DocStrings

Docstrings are used in **functions, classes, modules, and methods** to describe their purpose, behavior, parameters, and return values. They help developers understand what the code does without reading the full implementation.

They are **not like comments**. Comments are ignored by the interpreter at runtime, but docstrings are not ignored—they are stored in memory and attached to the object.

Docstrings can be accessed using tools like:

* `help(object)`
* `object.__doc__`

### Key Features:

* Written as a string literal (usually triple quotes `""" ... """`)
* Placed immediately after the definition of a function, class, or module
* Used for documentation and introspection
* Can be used by documentation generators (like Sphinx)

### Example Topics Covered in Docstrings:

* Function purpose
* Parameters (inputs)
* Return values (outputs)
* Class behavior and attributes
* Module overview

### Example (Google Style):

```python
def add(a, b):
    """
    Returns the sum of two numbers.

    Parameters:
        a (int or float): First number
        b (int or float): Second number

    Returns:
        int or float: Sum of a and b
    """
    return a + b
```

Docstrings are a key part of writing clean, readable, and maintainable Python code.


````python
def find_maximum(*args):
    """
    Find the maximum value from a set of numeric inputs.

    This function accepts multiple numeric values and returns the largest one.
    It validates that:
    - At least one value is provided
    - All values are int or float

    Parameters:
        *args: Variable number of numeric values (int or float)

    Returns:
        int or float: The maximum value among inputs
        None: If input is invalid or empty
    """
    try:
        if not args:
            raise ValueError("At least one number must be provided")

        for i, value in enumerate(args):
            print(f"Index: {i}, Value: {value}")
            if not isinstance(value, (int, float)):
                raise ValueError(f"Value at position {i} must be a number")

        return max(args)

    except ValueError as e:
        print(f"Error: {e}")
        return None

print(find_maximum(10, 5, 8, 20))
````

````python
print(f"Docstring for find_maximum:\n{find_maximum.__doc__}")
````

````python
print(f"Docstring for find_maximum:\n{help(find_maximum)}")
````

````python
class Calculator:
    """
    Simple calculator class for basic operations.
    """

    def square(self, x):
        """
        Return the square of a number.

        Parameters:
            x (int or float): Number to square

        Returns:
            int or float: Squared result
            None: If input is invalid
        """
        try:
            if not isinstance(x, (int, float)):
                raise ValueError("Input must be a number")

            return x * x

        except ValueError as e:
            print(f"Error: {e}")
            return None


calc = Calculator()
print(f"Docstring for Calculator.square:\n{Calculator.square.__doc__}")
````

# Lambda function

### What is a lambda function?

A **lambda function** is a small anonymous function in Python defined using the `lambda` keyword. It can take any number of inputs but contains only **one expression**.

```python
square = lambda x: x * x
print(square(5))
```


### Why use lambda functions?

### 1. For short and simple operations

Lambda is useful when the function is very small and used only once.

```python
add = lambda a, b: a + b
print(add(2, 3))
```

---

### 2. Used with built-in functions

Lambda works well with functions like `map()`, `filter()`, and `sorted()`.

```python
numbers = [1, 2, 3, 4]

result = list(map(lambda x: x * 2, numbers))
print(result)
```

---

### 3. Makes code shorter and cleaner

Instead of writing full functions, lambda helps reduce code length.

```python
# Normal function
def square(x):
    return x * x

# Lambda version
square = lambda x: x * x
```

---

### When NOT to use lambda

* If logic is complex
* If multiple lines are needed
* If code readability becomes poor

### Summary

* Lambda = small anonymous function
* Used for simple, quick operations
* Best with `map`, `filter`, `sorted`
* Not a replacement for full functions


````python
# Lambda function to calculate the square of a number
square_lambda = lambda x: x * x if isinstance(x, (int, float)) else None

print(f"Square using lambda: {square_lambda(5)}")
print(f"Square using lambda with invalid input: {square_lambda('a')}")
````

### Note (PEP 8 – `lambda` usage)

Using a `lambda` expression and assigning it to a variable is generally **not recommended** in Python style guidelines (PEP 8). This is because it reduces readability and goes against the intended purpose of `lambda`.

#### Avoid this pattern:

```python
add = lambda x, y: x + y
```

This is discouraged because:

* It makes the code less readable
* It hides the fact that a named function is being defined
* It violates common PEP 8 style expectations

#### Preferred approach:

Use a regular function definition instead:

```python
def add(x, y):
    return x + y
```

#### Proper use of `lambda`:

`lambda` should mainly be used for short, simple, anonymous functions—especially when passing a function as an argument to another function.

Example:

```python
numbers = [1, 2, 3, 4]
squared = list(map(lambda x: x * x, numbers))
```

Or:

```python
sorted_items = sorted(items, key=lambda item: item.price)
```

#### Summary:

* Use `def` for named functions
* Use `lambda` only for small, inline functions passed as arguments
* Avoid assigning `lambda` to variables for better readability and maintainability


### `sorted()` method (short detailed format)

```text
sorted(list, key=function, reverse=True/False)
```

### Meaning:

* **list** → the data you want to sort (any iterable)
* **key** → function (often `lambda`) that decides sorting logic
* **reverse (order)** → controls sorting direction

  * `False` → ascending order (default)
  * `True` → descending order

### Example:

```python id="s1t2u3"
sorted(numbers, key=lambda x: x, reverse=False)
```

### Summary:

* `key` = sorting rule
* `reverse` = order control
* `sorted()` = returns a new sorted list


````python
raw_list = [45, 2, 88, 31, 67, 90, 14, 59, 100, 23]

sorted_list = sorted(raw_list) # Default is ascending order
print(f"Sorted List: {sorted_list}")

sorted_list = sorted(raw_list, reverse=True)
print(f"Sorted List (Descending): {sorted_list}")
````

````python
sorted_numbers = sorted(raw_list, key=lambda x: x)
print(f"Sorted List using lambda: {sorted_numbers}")
````

````python
students = [("John", 85), ("Alice", 90), ("Bob", 78)]

sorted_students_by_name = sorted(students, key=lambda student: student[0], reverse=True)
print(f"Sorted Students by Name : {sorted_students_by_name}")

sorted_students = sorted(students, key=lambda student: student[1])
print(f"Sorted Students by Score: {sorted_students}")
````

````python
users = [
    {"name": "Alice", "age": 30},
    {"name": "Bob", "age": 25},
    {"name": "Charlie", "age": 35},
]
# 1. Sort users by age
print(f"Users sorted by age: {sorted(users, key=lambda user: user['age'])}")

# 2. Find user with maximum age
print(f"User with maximum age: {max(users, key=lambda user: user['age'])}")

# 3. Find user with minimum age
print(f"User with minimum age: {min(users, key=lambda user: user['age'])}")
````

# List Comprehensisons

A list comprehension is a concise way to create or filter lists.

### Basic Syntax

```python
[expression for item in iterable if condition]
```

### Filter Example

```python
numbers = [1, 2, 3, 4, 5, 6]

evens = [n for n in numbers if n % 2 == 0]

print(evens)
# Output: [2, 4, 6]
```

### Key Points

* Creates a new list.
* Can transform values using an expression.
* Can filter values using an `if` condition.
* Often more readable than loops for simple operations.


### Example

````python
numbers = [12, 45, 7, 23, 56, 89, 34, 2, 78, 91, 15, 67, 43, 28, 100, 5, 62, 39, 84, 11]

even_list = [number for number in numbers if number % 2 == 0]
print(f"Even Numbers: {even_list}")

odd_list = [number for number in numbers if number  % 2 != 0]
print(f"Odd Numbers: {odd_list}")

greater_than_5 = [number for number in numbers if number > 5]
print(f"Numbers greater than 5: {greater_than_5}")

less_than_30 = [number for number in numbers if number < 30]
print(f"Numbers less than 30: {less_than_30}")

squared_numbers = [number ** 2 for number in numbers]
print(f"Squared Numbers: {squared_numbers}")
````

# Dictionary Comprehensions

A dictionary comprehension is a concise way to create dictionaries.

### Basic Syntax

```python
{key_expression: value_expression for item in iterable}
```

### Example : Create a dictionary from a list

```python
labels = ["cat", "dog", "bird"]

label_dict = {label: index for index, label in enumerate(labels)}

print(label_dict)
```

Output:

```python
{'cat': 0, 'dog': 1, 'bird': 2}
```

### Key Points

* Creates a new dictionary.
* Can generate keys and values dynamically.
* Can include filtering using an `if` condition.
* Often replaces loops used to build dictionaries.
