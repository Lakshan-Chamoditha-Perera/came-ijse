# Inheritance

Inheritance allows a class (**subclass / child class**) to inherit attributes and methods from another class (**superclass / parent class**).

It promotes **code reusability** and helps create hierarchical relationships between classes.

## Basic Inheritance

```python
class Person:
    def __init__(self, name):
        self.name = name

    def display(self):
        print(f"Name: {self.name}")


class Student(Person):   # Student inherits Person
    def study(self):
        print(f"{self.name} is studying")


student = Student("Alice")
student.display()
student.study()
```

### Output
```text
Name: Alice
Alice is studying
```

---

## Superclass and Subclass Behavior

- **Superclass (Parent Class)** → Base class that provides common properties and methods.
- **Subclass (Child Class)** → Derived class that inherits from superclass and can add new features.

Example:
```python
class Animal:
    def sound(self):
        print("Animal makes sound")


class Dog(Animal):
    def bark(self):
        print("Dog barks")
```

Here:
- `Animal` → Superclass  
- `Dog` → Subclass  

---

## `__init__()` in Inheritance

When a subclass is created:
- If subclass has **no constructor**, parent constructor is automatically used.
- If subclass defines its own constructor, parent constructor is **not called automatically** unless using `super()`.

### Parent Constructor Automatically Used

```python
class Person:
    def __init__(self, name):
        self.name = name


class Student(Person):
    pass


student = Student("Alice")
print(student.name)
```

Output:
```text
Alice
```

---

## Using `super()` Keyword

`super()` is used to call methods or constructors of the parent class.

```python
class Person:
    def __init__(self, name):
        self.name = name


class Student(Person):
    def __init__(self, name, grade):
        super().__init__(name)   # Call parent constructor
        self.grade = grade

    def display(self):
        print(f"Name: {self.name}, Grade: {self.grade}")


student = Student("Alice", "A")
student.display()
```

Output:
```text
Name: Alice, Grade: A
```

### Why use `super()`?
- Reuse parent initialization logic
- Avoid duplicate code
- Access parent methods

## Using `super()` Keyword

`super()` is used to call the constructor or methods of the parent class.

### Without `super()`

Here, the subclass manually initializes parent attributes instead of calling the parent constructor.

```python
class Vehicle:
    def __init__(self, manufacturing_date, color):
        self.manufacturing_date = manufacturing_date
        self.color = color


class Car(Vehicle):
    def __init__(self, manufacturing_date, color, brand):
        # super().__init__(manufacturing_date, color)  # Parent constructor NOT called
        self.manufacturing_date = manufacturing_date
        self.color = color
        self.brand = brand

    def display(self):
        print(f"Manufacturing Date: {self.manufacturing_date}")
        print(f"Color: {self.color}")
        print(f"Brand: {self.brand}")


car = Car("2026-06-28", "Black", "BMW")
car.display()
```

### Output
```text
Manufacturing Date: 2026-06-28
Color: Black
Brand: BMW
```

### Problem Without `super()`
- Parent constructor is skipped
- Parent initialization logic is duplicated
- If parent constructor changes, subclass must also be updated

---

### With `super()`

```python
class Vehicle:
    def __init__(self, manufacturing_date, color):
        self.manufacturing_date = manufacturing_date
        self.color = color


class Car(Vehicle):
    def __init__(self, manufacturing_date, color, brand):
        super().__init__(manufacturing_date, color)  # Calls parent constructor
        self.brand = brand

    def display(self):
        print(f"Manufacturing Date: {self.manufacturing_date}")
        print(f"Color: {self.color}")
        print(f"Brand: {self.brand}")


car = Car("2026-06-28", "Black", "BMW")
car.display()
```

### Why `super()` is Better
- Reuses parent constructor code
- Avoids duplicate initialization
- Easier maintenance
- Cleaner object-oriented design

# Encapsulation 
Encapsulation is one of the fundamental concepts of Object-Oriented Programming (OOP). It refers to the bundling of data (attributes) and methods (functions) that operate on that data within a single unit (class), and restricting direct access to some of an object's components.

Think of encapsulation like a **capsule** - it hides the internal details and only exposes what's necessary. For example, when you take a medicine capsule, you don't need to know its chemical composition; you just need to know how to take it.

## Why Use Encapsulation?

1. **Data Protection**: Prevents accidental or unauthorized modification of data
2. **Simplification**: Hides complexity from the user
3. **Maintainability**: Internal implementation can be changed without affecting external code
4. **Control**: You can add validation logic when data is modified

---

## Access Modifiers in Python

Python has three types of access modifiers to control the visibility of class members:

| Modifier | Naming Convention | Access Level |
|----------|------------------|--------------|
| **Public** | `variable` | Accessible from anywhere |
| **Protected** | `_variable` | Accessible within class and subclasses |
| **Private** | `__variable` | Accessible only within the class |


### 1. Public Members

Public members are accessible from anywhere - inside the class, outside the class, and in subclasses. **This is the default access level in Python.**


````python
class Student:
    def __init__(self, name, age):
        self.name = name  # Public attribute
        self.age = age  # Public attribute

    def display_info(self):  # Public method
        print(f"Name: {self.name}, Age: {self.age}")


# Using public members
student1 = Student("Alice", 20)

# Accessing from outside the class
print(student1.name)  # Output: Alice
print(student1.age)  # Output: 20

# Modifying from outside
student1.name = "Bob"
student1.age = 25

# Calling public method
student1.display_info()  # Output: Name: Bob, Age: 25
````

### 2. Protected Members

Protected members are prefixed with a single underscore (`_`).

They are **meant for internal use within the class and subclasses**, but Python does **not strictly prevent access from outside** (it’s a convention).

### Key Points
- `_variable` → Protected attribute  
- `_method()` → Protected method  
- Accessible inside:
  - Same class
  - Child/subclasses  
- Outside access is **possible but discouraged** (Python relies on developer discipline)

````python
class Student:
    def __init__(self, name, age):
        self._name = name  # Protected attribute
        self._age = age  # Protected attribute

    def _display_info(self):  # Protected method
        print(f"Name: {self._name}, Age: {self._age}")


# Child class accessing protected members
class GraduateStudent(Student):
    def show_details(self):
        print(f"Graduate Student: {self._name}, Age: {self._age}")
        self._display_info()


# Using protected members
student1 = GraduateStudent("Alice", 20)

# Accessing protected members inside subclass
student1.show_details()

# Accessing from outside class (possible, but not recommended)
print(student1._name)  # Output: Alice
print(student1._age)  # Output: 20
````

````python
class Vehicle:
    def __init__(self, manufacturing_date, color):
        self._manufacturing_date = manufacturing_date  # Protected attribute
        self._color = color  # Protected attribute

    def _print_manufacturing_date(self):  # Protected method
        print(f"Manufacturing Date (from vehicle): {self._manufacturing_date}")

    def _print_color(self):  # Protected method
        print(f"Color (from vehicle): {self._color}")


class Car(Vehicle):
    def __init__(self, manufacturing_date, color, brand):
        super().__init__(
            manufacturing_date, color
        )  # Call the constructor of the parent class
        self._brand = brand


car = Car("2026-06-28", "Black", "BMW")

print(
    car._manufacturing_date
)  # Accessing protected attribute from outside the class (not recommended)
print(
    car._color
)  # Accessing protected attribute from outside the class (not recommended)


car._print_manufacturing_date()  # Calling protected method to display information
car._print_color()  # Calling protected method to display information
````

### 3. Private Members

Private members are prefixed with double underscores (`__`).

They are **intended to be accessible only within the class**. Python uses **name mangling** to make direct access from outside more difficult.

### Key Points
- `__variable` → Private attribute  
- `__method()` → Private method  
- Accessible inside:
  - Same class only  
- Outside direct access is **not allowed** (Python uses name mangling)
```

````python
class Student:
    def __init__(self, name, age):
        self.__name = name  # Private attribute
        self.__age = age  # Private attribute

    def __display_info(self):  # Private method
        print(f"Name: {self.__name}, Age: {self.__age}")

    def show_info(self):  # Public method to access private method
        self.__display_info()


# Using private members
student1 = Student("Alice", 20)

# Accessing private members through public method
student1.show_info()

# Direct access from outside class (raises error)
# print(student1.__name)   # AttributeError
# print(student1.__age)    # AttributeError
````

````python
class Vehicle:
    def __init__(self, manufacturing_date, color):
        self.__manufacturing_date = manufacturing_date
        self.__color = color

    def __display(self):
        print(
            f"Vehicle : __manufacturing_date: {self.manufacturing_date} color: {self.__color}"
        )


class Car(Vehicle):
    def __init__(self, manufacturing_date, color):
        super().__init__(manufacturing_date, color)


car = Car("2026-06-05", "Black")
# print(car.__color) // AttributeError
# car.__display() // AttributeError
````

````python
class Vehicle:
    def __init__(self, manufacturing_date, color):
        self.__manufacturing_date = manufacturing_date
        self.__color = color

    # Getter for color (read-only property)
    @property
    def color(self):
        return self.__color

    # Getter for manufacturing_date (read-only property)
    @property
    def manufacturing_date(self):
        return self.__manufacturing_date

    # Getter for info
    @property
    def display(self):
        print(
            f"Vehicle : __manufacturing_date: {self.manufacturing_date} color: {self.color}"
        )


class Car(Vehicle):
    def __init__(self, manufacturing_date, color):
        super().__init__(manufacturing_date, color)


car = Car("2026-06-05", "Black")

print("Accessing private attributes using getters:")
print(f"Color : {car.color}")  # Accessing color using getter
print(
    f"Manufacturing Date : {car.manufacturing_date}"
)  # Accessing manufacturing_date using getter

car.display  # Calling display method to show information
````

# Composition (HAS-A Relationship)


Composition is a core concept in Object-Oriented Programming (OOP) where a class is built using one or more objects of other classes.

Instead of inheriting behavior (IS-A relationship), composition focuses on **using objects inside other objects (HAS-A relationship)**.

Example: A Car has an Engine and a Battery.

## Key Idea

- Inheritance → IS-A relationship  
- Composition → HAS-A relationship  

Composition helps you design systems by combining independent components.


## Why Use Composition?

- Code Reusability → reuse existing classes  
- Modularity → break system into small parts  
- Loose Coupling → classes are independent  
- Flexibility → swap components easily  
- Better Design → closer to real-world systems  

## HAS-A Relationship Example

- A `Car` HAS an `Engine`  
- A `Car` HAS a `Battery`  


````python
class Engine:
    def start(self):
        print("Engine started...")


class Battery:
    def charge(self):
        print("Battery charging...")


class Car:
    def __init__(self):
        self.engine = Engine()
        self.battery = Battery()

    def start_car(self):
        print("Starting car...")
        self.engine.start()

    def charge_car(self):
        print("Charging car...")
        self.battery.charge()


car = Car()

car.start_car()
car.charge_car()
````

---

````python
class Battery:
    def __init__(self, charging_capacity):
        # Private attribute to store battery capacity (kWh)
        self.__charging_capacity = charging_capacity

    @property
    def capacity(self):
        # Read-only property to access battery capacity safely
        return self.__charging_capacity

    def charge(self):
        # Simulates charging the battery
        print(f"Battery with capacity {self.__charging_capacity} kWh is charging...")


class Engine:
    def __init__(self, horse_power):
        # Private attribute to store engine power
        self.__horse_power = horse_power

    @property
    def horse_power(self):
        # Read-only property to access engine horsepower
        return self.__horse_power

    def start(self):
        # Simulates engine starting process
        print(f"{self.__horse_power} HP engine started...")


class Car:
    def __init__(self, model, horse_power, charging_capacity):
        # Private attribute for car model
        self.__model = model

        # Composition (HAS-A relationship)
        # Car HAS an Engine
        self.__engine = Engine(horse_power)

        # Car HAS a Battery
        self.__battery = Battery(charging_capacity)

    @property
    def model(self):
        # Read-only property for car model
        return self.__model

    def start_car(self):
        # Start the car by starting its engine
        print(f"Car Model: {self.__model}")
        self.__engine.start()

    def charge_car(self):
        # Charge the car battery using Battery class
        print(f"Charging Capacity: {self.__battery.capacity} kWh")
        self.__battery.charge()
````

````python
car = Car(model="Tesla Model S", horse_power=670, charging_capacity=100)
car.start_car()
car.charge_car()
````
