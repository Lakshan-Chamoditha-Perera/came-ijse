# Variables and Naming

A **variable** is a named container for storing data. Think of it like a labelled box — you put something inside it, and you can retrieve or change it later using its name.

```
  name = "Alice"
  ┌─────────┐
  │  "Alice" │  ← stored value
  └─────────┘
      ↑
    "name"     ← variable label
```

## 3.1 Naming Rules and Conventions

Python has **strict rules** (must follow) and **conventions** (should follow for readability).

### 📏 Hard Rules (your code will crash if broken)

| Rule | Valid ✅ | Invalid ❌ |
|------|---------|----------|
| Start with a letter or underscore | `name`, `_value` | `1name`, `@value` |
| Only letters, digits, underscores | `my_var2` | `my-var`, `my var` |
| Case-sensitive | `age` ≠ `Age` ≠ `AGE` | — |
| Cannot be a Python keyword | `for_loop` | `for`, `if`, `class` |

### 🐍 Python Reserved Keywords (cannot use as variable names)

```
False   None    True    and     as      assert  async   await
break   class   continue def    del     elif    else    except
finally for     from    global  if      import  in      is
lambda  nonlocal not     or      pass    raise   return  try
while   with    yield
```

### 🎨 Conventions (PEP 8 Style Guide — Python's official style)

| Convention | Example | Used For |
|-----------|---------|----------|
| `snake_case` | `user_name`, `total_price` | Variables, functions |
| `SCREAMING_SNAKE_CASE` | `MAX_SIZE`, `PI` | Constants |
| `PascalCase` | `MyClass`, `UserProfile` | Class names |
| `_leading_underscore` | `_private_var` | "Private" variables |
| Descriptive names | `user_age` not `ua` | Always! |

````python
# ✅ Valid variable names
name = "Alice"
user_age = 25
_hidden = "secret"
totalPrice2024 = 199.99
MAX_RETRIES = 3

print(name)
print(user_age)
print(_hidden)
print(totalPrice2024)
print(MAX_RETRIES)
````

````python
# Python is CASE-SENSITIVE — these are THREE different variables!
score = 10
Score = 20
SCORE = 30

print(score)   # 10
print(Score)   # 20
print(SCORE)   # 30
````

````python
# Assigning and reassigning variables
x = 5
print("x is:", x)

x = 10          # Reassign — the old value is gone
print("x is now:", x)

x = x + 1      # Use x's current value to compute a new value
print("x after +1:", x)

# Multiple assignment on one line
a, b, c = 1, 2, 3
print(a, b, c)

# Assign the same value to multiple variables
x = y = z = 0
print(x, y, z)
````

## 4.1 Strings (`str`)

A **string** is a sequence of characters — letters, numbers, symbols, spaces. Strings represent text.

Strings are enclosed in:
- Single quotes: `'hello'`
- Double quotes: `"hello"`
- Triple quotes: `'''hello'''` or `"""hello"""` (for multiline text)

````python
# Creating strings — single and double quotes are interchangeable
first_name = 'Alice'
last_name = "Smith"
greeting = "Hello, World!"

print(first_name)
print(last_name)
print(greeting)
print(type(greeting))  # <class 'str'>
````

````python
# Use double quotes when your string contains an apostrophe
sentence = "It's a beautiful day!"
print(sentence)

# Use single quotes when your string contains double quotes
quote = 'She said "Python is great!"'
print(quote)

# Or escape the quote with a backslash \
escaped = 'It\'s also fine like this'
print(escaped)
````

````python
# Triple-quoted strings span multiple lines
poem = """Roses are red,
Violets are blue,
Python is awesome,
And so are you!"""

print(poem)
````

````python
# --- String Operations ---

# Concatenation: joining strings with +
full_name = first_name + " " + last_name
print("Full name:", full_name)

# Repetition: repeating with *
border = "-" * 20
print(border)

# Length: how many characters?
print("Length of greeting:", len(greeting))

# Indexing: get a single character (0-indexed!)
word = "Python"
print("First character:", word[0])   # P
print("Last character:", word[-1])   # n

# Slicing: get a portion of the string [start:end]
print("First 3 chars:", word[0:3])   # Pyt
print("Last 3 chars:", word[-3:])    # hon
````

````python
# --- Useful String Methods ---
text = "  Hello, Python World!  "

print(text.upper())          # ALL CAPS
print(text.lower())          # all lowercase
print(text.strip())          # remove whitespace from both ends
print(text.strip().replace("Python", "Amazing"))  # replace a word
print(text.strip().split(", "))   # split into a list at ", "
print(text.strip().startswith("Hello"))  # check start
print(text.strip().endswith("!"))        # check end
print("Python" in text)      # check if substring exists → True/False
````

````python
# --- f-strings (formatted strings) — the modern way to embed variables ---
# Prefix the string with f and use {variable} inside

name = "Alice"
age = 25
city = "Colombo"

# Old way (concatenation — messy)
print("My name is " + name + " and I am " + str(age) + " years old.")

# f-string way (clean and readable)
print(f"My name is {name} and I am {age} years old.")
print(f"{name} lives in {city}.")

# You can even put expressions inside {}
price = 49.99
quantity = 3
print(f"Total cost: ${price * quantity:.2f}")  # .2f = 2 decimal places
````

## 4.2 Integers and Floats (`int` and `float`)

Python has two main numeric types:

| Type | Description | Example |
|------|-------------|--------|
| `int` | Whole numbers (no decimal point) | `0`, `42`, `-7`, `1_000_000` |
| `float` | Numbers with a decimal point | `3.14`, `-0.5`, `2.0`, `1.5e10` |

````python
# --- Integers ---
age = 25
year = 2024
negative = -100
big_number = 1_000_000   # Underscores improve readability (Python 3.6+)

print(age, type(age))
print(year, type(year))
print(negative, type(negative))
print(big_number, type(big_number))
````

````python
# --- Floats ---
pi = 3.14159
temperature = -2.5
price = 19.99
whole_as_float = 2.0   # Note: 2.0 is a float, 2 is an int

print(pi, type(pi))
print(temperature, type(temperature))
print(whole_as_float, type(whole_as_float))
````

````python
# --- Arithmetic Operators ---
a = 17
b = 5

print(f"{a} + {b}  = {a + b}")    # Addition
print(f"{a} - {b}  = {a - b}")    # Subtraction
print(f"{a} * {b}  = {a * b}")    # Multiplication
print(f"{a} / {b}  = {a / b}")    # Division       → always returns float!
print(f"{a} // {b} = {a // b}")   # Floor division → integer result
print(f"{a} % {b}  = {a % b}")    # Modulo (remainder)
print(f"{a} ** {b} = {a ** b}")   # Exponentiation (17 to the power of 5)
````

````python
# --- Type Conversion (Casting) ---

# int → float
x = 5
print(float(x))       # 5.0

# float → int (truncates — cuts off the decimal, does NOT round!)
y = 9.99
print(int(y))         # 9  (not 10!)

# string → int (only works if the string looks like a number)
num_str = "42"
print(int(num_str))   # 42
print(type(int(num_str)))  # int

# int → string
print(str(100))       # "100"
````

````python
# ⚠️ FLOAT CAUTION: Floating-Point Precision Issues
# 
# Computers store floats in binary (base 2).
# Most decimal fractions CANNOT be represented exactly in binary.
# This causes tiny rounding errors that can surprise you!

print(0.1 + 0.2)          # You might expect 0.3...
print(0.1 + 0.2 == 0.3)   # Is it exactly 0.3? Let's check...
print(repr(0.1 + 0.2))    # See the full precision
````

````python
# More float precision surprises
print(1.1 + 2.2)           # Not exactly 3.3!
print(0.1 * 3)             # Not exactly 0.3
print(0.3 - 0.2 - 0.1)    # Should be 0.0...
print()

# ✅ SOLUTION 1: round() for display
result = 0.1 + 0.2
print(round(result, 2))    # 0.3 — rounded to 2 decimal places

# ✅ SOLUTION 2: Use round() for comparisons too
print(round(0.1 + 0.2, 10) == round(0.3, 10))  # True

# ✅ SOLUTION 3: Use math.isclose() for safe float comparison
import math
print(math.isclose(0.1 + 0.2, 0.3))   # True — the proper way!

# ✅ SOLUTION 4: Use the decimal module for financial/precise calculations
from decimal import Decimal
print(Decimal('0.1') + Decimal('0.2'))  # Exactly 0.3
````

````python
# --- Scientific Notation (e-notation) ---
# Very large or very small numbers can be written with 'e'
# 'e' means "times 10 to the power of"

# Large numbers
speed_of_light = 3e8       # 3 × 10^8  = 300,000,000 m/s
avogadro = 6.022e23        # 6.022 × 10^23

# Small numbers
electron_charge = 1.6e-19  # 1.6 × 10^-19 coulombs
planck = 6.626e-34         # Planck's constant

print(f"Speed of light:    {speed_of_light}")
print(f"Avogadro's number: {avogadro}")
print(f"Electron charge:   {electron_charge}")
print(f"Planck's constant: {planck}")
print()

# Scientific notation in arithmetic
print(f"2.5e3 = {2.5e3}")     # 2500.0
print(f"5e-2 = {5e-2}")       # 0.05
print(f"1e6 * 1e6 = {1e6 * 1e6}")  # 1 trillion
````

````python
# --- Special Float Values ---
import math

# Infinity
pos_inf = float('inf')
neg_inf = float('-inf')
print(pos_inf)                 # inf
print(neg_inf)                 # -inf
print(pos_inf > 1_000_000)     # True — infinity is bigger than anything

# NaN — Not a Number (result of undefined operations)
nan_val = float('nan')
print(nan_val)                 # nan
print(math.isnan(nan_val))     # True — use isnan() to check for NaN
print(nan_val == nan_val)      # False! NaN is not equal to itself!
````

## 4.3 Boolean (`bool`)

A **Boolean** represents one of exactly two values: `True` or `False`. Booleans are the foundation of all logic and decision-making in programs.

> ⚠️ **Important:** In Python, `True` and `False` are **capitalized**. `true` or `false` (lowercase) will cause a `NameError`!

````python
# --- Boolean Basics ---
is_raining = True
is_sunny = False
is_adult = True

print(is_raining)
print(is_sunny)
print(type(is_raining))   # <class 'bool'>
````

````python
# --- Comparison Operators → always return a Boolean ---
a = 10
b = 5

print(f"{a} == {b}  → {a == b}")   # Equal to
print(f"{a} != {b}  → {a != b}")   # Not equal to
print(f"{a} >  {b}  → {a > b}")    # Greater than
print(f"{a} <  {b}  → {a < b}")    # Less than
print(f"{a} >= {b}  → {a >= b}")   # Greater than or equal to
print(f"{a} <= {b}  → {a <= b}")   # Less than or equal to
````

````python
# --- Logical Operators: and, or, not ---

x = True
y = False

# AND: both must be True
print(f"True  and True  → {True and True}")    # True
print(f"True  and False → {True and False}")   # False
print(f"False and False → {False and False}")  # False

print()

# OR: at least one must be True
print(f"True  or True   → {True or True}")     # True
print(f"True  or False  → {True or False}")    # True
print(f"False or False  → {False or False}")   # False

print()

# NOT: flips the value
print(f"not True   → {not True}")   # False
print(f"not False  → {not False}")  # True
````

````python
# --- Real-world Boolean logic ---
age = 20
has_id = True
is_member = False

# Can enter a venue?
can_enter = age >= 18 and has_id
print(f"Can enter venue: {can_enter}")

# Gets a discount?
gets_discount = age < 12 or is_member
print(f"Gets discount: {gets_discount}")

# Is a non-member?
is_non_member = not is_member
print(f"Is non-member: {is_non_member}")

# Combined condition
premium_access = age >= 18 and (is_member or has_id)
print(f"Premium access: {premium_access}")
````

````python
# --- Truthiness: Every Python value has a boolean meaning ---
# bool() converts any value to True or False

print("=== FALSY values (evaluate to False) ===")
print(bool(0))        # False — zero integer
print(bool(0.0))      # False — zero float
print(bool(""))       # False — empty string
print(bool(None))     # False — None
print(bool([]))       # False — empty list

print()
print("=== TRUTHY values (evaluate to True) ===")
print(bool(1))        # True
print(bool(-5))       # True — any non-zero number
print(bool("hi"))     # True — any non-empty string
print(bool([1, 2]))   # True — non-empty list
````

````python
# --- Booleans are actually integers in Python! ---
# True == 1, False == 0

print(True + True)       # 2
print(True + False)      # 1
print(False + False)     # 0
print(True * 10)         # 10

# Useful: count how many True values are in a list
results = [True, False, True, True, False]
print(f"Number of True values: {sum(results)}")  # 3
````

## 4.4 None (`NoneType`)

`None` is Python's way of representing **the absence of a value** — "nothing", "empty", "no result".

It is its own special type called `NoneType`, and there is **only one** `None` object in all of Python.

### When does `None` appear?
- A variable that hasn't been given a meaningful value yet
- A function that doesn't explicitly return anything
- Optional function parameters not provided
- Indicating "no result found" or "missing data"

> 💡 `None` is **not** the same as `0`, `False`, or `""`. Those are actual values. `None` means *there is no value*.

````python
# --- None basics ---
result = None
middle_name = None   # Person might not have a middle name
winner = None        # No winner yet

print(result)
print(middle_name)
print(type(None))    # <class 'NoneType'>
````

````python
# --- Checking for None ---
# ALWAYS use 'is' or 'is not' to check for None, NOT == or !=

value = None

# ✅ Correct way
print(value is None)       # True
print(value is not None)   # False

# ⚠️ This works but is discouraged (== can be overridden in classes)
print(value == None)       # True (works, but not recommended)

# Practical usage in an if-statement
user_input = None

if user_input is None:
    print("No input provided yet.")
else:
    print(f"User said: {user_input}")
````

````python
# --- Functions that return None ---
# A function without a 'return' statement implicitly returns None

def greet(name):
    print(f"Hello, {name}!")   # This prints but doesn't return a value

output = greet("Alice")   # The function runs and prints...
print(output)             # ...but 'output' is None!
print(output is None)     # True
````

````python
# --- None vs 0 vs False vs "" — they are NOT the same! ---

print("=== Equality checks ===")
print(None == 0)       # False — None is not zero
print(None == False)   # False — None is not False
print(None == "")      # False — None is not an empty string

print()
print("=== But all are falsy ===")
print(bool(None))      # False — None is falsy
print(bool(0))         # False — 0 is falsy
print(bool(False))     # False
print(bool(""))        # False — empty string is falsy

print()
# Think of it this way:
# 0    means "I counted and got zero"
# False means "the answer is no"
# ""   means "I have an empty text"
# None means "I have no answer at all"
print("None = no value at all")
print("0    = the value zero")
print('""   = an empty string (a value that exists, but is empty)')
````

````python
# --- Practical None patterns ---

# Pattern 1: Initialize a variable that will be assigned later
best_score = None
scores = [88, 72, 95, 61, 99]

for score in scores:
    if best_score is None or score > best_score:
        best_score = score

print(f"Best score: {best_score}")

# Pattern 2: Optional parameter
def create_profile(name, age, email=None):
    profile = f"Name: {name}, Age: {age}"
    if email is not None:
        profile += f", Email: {email}"
    return profile

print(create_profile("Bob", 30))
print(create_profile("Alice", 25, "alice@example.com"))
````
