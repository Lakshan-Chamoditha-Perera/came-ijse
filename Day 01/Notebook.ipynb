# 3. Variables and Naming

In Java you declare a variable like this:
```java
String name = "Alice";
int age = 25;
```
In Python there is **no type declaration** and **no semicolon**.
You just write the name and assign a value:
```python
name = "Alice"
age = 25
```
Python figures out the type automatically. This is called **dynamic typing**.

⭐ **Remember:** Python variables are not typed — the *value* has a type, not the variable.

````python
# No 'int', 'String', or 'double' keyword needed.
# No semicolon at the end of the line.
name = "Alice"
age  = 25
gpa  = 3.85

# type() tells you what data type a value is — useful for debugging
print(type(name))   # <class 'str'>
print(type(age))    # <class 'int'>
print(type(gpa))    # <class 'float'>
````

````python
# ⭐ In Python you can reassign a variable to a DIFFERENT type.
# Java would not allow this — but Python allows it.
x = 10          # x is an int
print(type(x))  # <class 'int'>

x = "hello"     # now x is a str — completely valid in Python
print(type(x))  # <class 'str'>
````

````python
# Python lets you assign multiple variables in ONE line
a, b, c = 1, 2, 3
print(a, b, c)  # 1 2 3

# You can also assign the same value to multiple variables at once
x = y = z = 0
print(x, y, z)  # 0 0 0
````

## 3.1 Naming Rules and Conventions

### ✅ Rules (enforced by Python — breaking these causes an error)

| Rule | Example |
|---|---|
| Must start with a letter or underscore `_` | `name`, `_name` ✅ |
| Cannot start with a number | `1name` ❌ |
| Can contain letters, numbers, underscores | `my_var1` ✅ |
| Cannot contain spaces or special characters | `my-var`, `my var` ❌ |
| Case-sensitive | `Name` and `name` are two different variables |
| Cannot use reserved keywords | `if`, `for`, `class`, `True`, `None` etc. ❌ |

### 🎨 Conventions (not enforced — but every Python developer follows these)

| Style | Used for | Example |
|---|---|---|
| `snake_case` | variables and functions | `user_name`, `get_data()` |
| `PascalCase` | classes | `StudentRecord` |
| `UPPER_SNAKE_CASE` | constants | `MAX_SIZE = 100` |
| Single underscore `_` | throwaway variable | `for _ in range(3):` |
| Double underscore `__name` | special Python internals | `__init__`, `__str__` |

⭐ **Java vs Python naming:**  
Java uses `camelCase` for variables → `studentName`  
Python uses `snake_case` for variables → `student_name`

````python
# ✅ GOOD names — follow Python snake_case convention
student_name   = "Alice"
total_score    = 95
is_enrolled    = True
MAX_STUDENTS   = 30      # constant — use UPPER_SNAKE_CASE

# ❌ These would cause a SyntaxError — uncomment to see the error
# 1student = "Alice"    # starts with a number
# student-name = "Alice" # hyphen not allowed
# class = "Biology"     # 'class' is a reserved keyword

print(student_name, total_score, is_enrolled, MAX_STUDENTS)
````

## 4.1 Strings (`str`)

A string is a sequence of characters. In Java, `String` is a class. In Python, `str` is a built-in type — simpler and more powerful.

⭐ **Key differences from Java:**
- Use single `'...'` OR double `"..."` quotes — both are identical in Python
- Triple quotes `"""..."""` for multi-line strings (no `\n` needed)
- Strings are **immutable** just like Java — methods return new strings
- No `char` type in Python — a single character is just a one-letter string

````python
# Three ways to create a string — all are valid
single   = 'Hello'
double   = "Hello"
multiline = """This is
a multi-line
string."""

print(single)
print(double)
print(multiline)
````

````python
# ⭐ f-strings: the modern way to embed variables inside strings
# Java equivalent: String.format() or + concatenation
name = "Alice"
age  = 25

# Old way (like Java's + concatenation) — avoid this in Python
print("Name: " + name + ", Age: " + str(age))

# Modern way — f-string (prefix the string with f, then use {} )
print(f"Name: {name}, Age: {age}")

# You can even put expressions inside {}
print(f"Next year Alice will be {age + 1}")
````

````python
# Strings are sequences — you can access characters by index
# ⭐ Python supports NEGATIVE indexing — Java does not
word = "Python"
#        P  y  t  h  o  n
# index  0  1  2  3  4  5
# neg   -6 -5 -4 -3 -2 -1

print(word[0])    # P   — first character
print(word[-1])   # n   — last character (negative index counts from end)
print(word[1:4])  # yth — slicing: characters from index 1 up to (not including) 4
````

````python
# Common string methods
sentence = "     lorem ipsum  dolor sit amet. Day 1.ipynb       "

print("len         :", len(sentence))                        # character count
print("upper       :", sentence.upper())                     # ALL CAPS
print("lower       :", sentence.lower())                     # all lowercase
print("strip       :", sentence.strip())                     # remove edge spaces
print("startswith  :", sentence.startswith("Day"))           # False — starts with spaces
print("endswith    :", sentence.endswith(".ipynb"))          # False — ends with spaces
print("'1' in s    :", "1" in sentence)                     # True — found anywhere
print("split '.'   :", sentence.split("."))                 # list of parts
print("replace     :", sentence.replace("Day", "Night"))    # swap words
````

````python
# ⭐ Strings are IMMUTABLE — you cannot change a character in place
# This is the same as Java
word = "Python"

# This will raise a TypeError — uncomment to see
# word[0] = "J"   # TypeError: 'str' object does not support item assignment

# Instead, create a new string
word = "J" + word[1:]
print(word)  # Jython
````

## 4.2 Integers (`int`) and Floats (`float`)

⭐ **Big difference from Java:**
- Java has `int` (32-bit), `long` (64-bit) — they have a max size
- Python `int` has **no size limit** — it grows as large as your memory allows
- No `long`, no `Integer.MAX_VALUE` overflow errors in Python

````python
# Regular integer
small = 42
print(small, type(small))

# ⭐ Python int has NO upper limit — try this in Java and it overflows!
huge = 99999999999999999999999999999999999999999999999999
print(huge)
print(type(huge))  # still just <class 'int'>

# Underscores in numbers — for readability (Python 3.6+)
# Java also supports this: int salary = 1_000_000;
salary = 1_000_000
print(salary)  # 1000000
````

````python
# float — decimal numbers (like Java's double)
pi = 3.14159
print(pi, type(pi))

# ⭐ Integer division in Python 3 uses // (floor division)
# In Java: 7 / 2 = 3 (integer division)
# In Python: 7 / 2 = 3.5 (always float!)
print(7 / 2)    # 3.5  — true division, always returns float
print(7 // 2)   # 3    — floor division (same as Java's int division)
print(7 % 2)    # 1    — modulo (remainder), same as Java
print(2 ** 10)  # 1024 — power operator (Java uses Math.pow)
````

````python
# Type conversion (casting)
# Java: (int) 3.9  →  Python: int(3.9)

x = int(3.9)       # truncates decimal — result is 3, NOT 4
y = float(5)       # 5 becomes 5.0
z = str(42)        # number becomes string '42'
n = int("100")    # string '100' becomes integer 100

print(x, type(x))
print(y, type(y))
print(z, type(z))
print(n, type(n))
````

## 4.3 Boolean (`bool`)

Booleans represent `True` or `False`.

⭐ **Key differences from Java:**
- Java: `true` / `false` (lowercase)
- Python: `True` / `False` — **capital first letter is required**
- Python `bool` is actually a subclass of `int` — `True == 1` and `False == 0`
- Python has a concept of **truthy** and **falsy** values — many things can act like a boolean

````python
# Boolean values — capital T and F are required
is_student  = True
is_graduate = False

print(is_student,  type(is_student))
print(is_graduate, type(is_graduate))

# ⭐ true and false (lowercase) are NOT booleans — they would be variable names
# This would cause a NameError:
# x = true   # NameError: name 'true' is not defined
````

````python
# Comparison operators — return True or False
# These are the same as Java
print(5 > 3)    # True
print(5 < 3)    # False
print(5 == 5)   # True  — equality (== not =)
print(5 != 3)   # True  — not equal
print(5 >= 5)   # True
print(5 <= 4)   # False
````

````python
# Logical operators
# ⭐ Python uses WORDS: 'and', 'or', 'not'
# Java uses symbols: &&, ||, !

x = True
y = False

print(x and y)   # False  (Java: x && y)
print(x or y)    # True   (Java: x || y)
print(not x)     # False  (Java: !x)

# Practical example
age       = 20
has_id    = True
can_enter = age >= 18 and has_id
print("Can enter:", can_enter)  # True
````

````python
# ⭐ In Python, bool is a subclass of int
# True == 1 and False == 0  (Java does NOT allow this)
print(True  == 1)   # True
print(False == 0)   # True
print(True  + True) # 2  — you can do arithmetic on booleans!
print(int(True))    # 1
print(int(False))   # 0
````

````python
# ⭐ Truthy and Falsy — unique to Python (Java does not have this)
# In Python, any value can be used in a boolean context (if statement, while, etc.)

# FALSY values — these all behave like False:
falsy_examples = [0, 0.0, "", [], {}, None, False]

# TRUTHY values — everything else behaves like True
truthy_examples = [1, -1, 3.14, "hello", [1, 2], {"a": 1}, True]

print("--- Falsy values ---")
for val in falsy_examples:
    print(f"  bool({str(val):<10}) = {bool(val)}")

print("\n--- Truthy values ---")
for val in truthy_examples:
    print(f"  bool({str(val):<10}) = {bool(val)}")
````

## 4.4 None (`NoneType`)

`None` is Python's way of saying "no value" or "empty".

⭐ **Java equivalent:** `null`

⭐ **Key differences from Java's `null`:**
- Java `null` can cause a `NullPointerException` — Python's `None` is safer to work with
- `None` is a real object in Python — it has a type (`NoneType`)
- There is only ONE `None` object in all of Python — always use `is None` to check, not `== None`

````python
# None represents 'no value' — equivalent to Java's null
result = None

print(result)           # None
print(type(result))     # <class 'NoneType'>
````

````python
# ⭐ Always check for None using 'is' — not '=='
# Java equivalent: if (result == null)
# Python correct:  if result is None

result = None

# ✅ Correct way
if result is None:
    print("result has no value")

# ✅ Check for NOT None
if result is not None:
    print("result has a value")
else:
    print("result is still None")

# ⚠️ This works but is not recommended
# if result == None:  <-- avoid this style
````

````python
# ⭐ Functions that don't return a value implicitly return None
# Java equivalent: a void method returns nothing

def say_hello():
    print("Hello!")  # no return statement

result = say_hello()    # prints Hello!
print(result)           # None — because the function returned nothing
print(type(result))     # <class 'NoneType'>
````

````python
# ⭐ None is falsy — it behaves like False in boolean context
value = None

if value:
    print("has a value")
else:
    print("value is None or empty")  # this will print

# Useful pattern: provide a default when value is None
name = None
display_name = name if name is not None else "Guest"
print(display_name)  # Guest
````

# 📋 Summary — Java vs Python Quick Reference

| Concept | Java | Python |
|---|---|---|
| Declare variable | `int x = 5;` | `x = 5` |
| Naming style | `camelCase` | `snake_case` |
| Integer limit | `int` max ~2.1 billion | **No limit** |
| True/False | `true` / `false` | `True` / `False` |
| Null value | `null` | `None` |
| Null check | `x == null` | `x is None` ✅ |
| Logical AND | `&&` | `and` |
| Logical OR | `\|\|` | `or` |
| Logical NOT | `!` | `not` |
| Integer division | `7 / 2 = 3` | `7 // 2 = 3` |
| True division | cast needed | `7 / 2 = 3.5` (default) |
| Power | `Math.pow(2, 10)` | `2 ** 10` |
| String format | `String.format()` | `f"Hello {name}"` |
| Type check | `instanceof` | `type(x)` or `isinstance(x, int)` |
| No return value | `void` | implicit `None` |

---
⭐ **Top things to remember from Java background:**
1. No type declarations — Python figures it out
2. `True` / `False` — capital first letter (not `true`/`false`)
3. `None` not `null` — and always check with `is None`
4. `int` has no size limit in Python
5. `/` always gives a float — use `//` for integer division
6. `and`, `or`, `not` — not `&&`, `||`, `!`
7. No semicolons, no curly braces — Python uses indentation
