````python
import numpy as np
````

````python
list = [1, 2, 3, 4, 5]
numpy_array = np.array(list)

print("List:", list)
print("Numpy Array:", numpy_array)

print ("Type of list:", type(list))
print ("Type of numpy array:", type(numpy_array))
````

Numpy arrays support element-wise operations, which means that you can perform operations on entire arrays without the need for explicit loops.
This makes Numpy arrays more efficient and faster for numerical computations compared to Python lists. 

````python
array_1 = np.array([10, 20, 30, 40, 50])
array_2 = np.array([10, 10, 10, 10, 10])

print(f"Addition       : {array_1 + array_2}") 
print(f"Subtraction    : {array_1 - array_2}") 
print(f"Multiplication : {array_1 * array_2}")
print(f"Division       : {array_1 / array_2}")
````

````python
array_a = np.array([1, 2, 3])
array_b = np.array([1, 1, 1])

def dot_product(a, b):
    if len(a) != len(b):
        raise ValueError("Arrays must be of the same length")
    return sum(a * b)     

print("Dot Product Result:", dot_product(array_a, array_b))
````

Assume we have a 2D NumPy array:

```python
import numpy as np

customers = np.array([
    ["Alice", 25],
    ["Bob", 30],
    ["Charlie", 28]
])
```

### Printing the Array

```python
print("Customer Data:\n", customers)
```

* Displays the entire 2D array.
* `\n` starts the array on a new line for better readability.

---

### `.shape`

```python
print("Customers shape:", customers.shape)
```

* Returns the dimensions of the array as **(rows, columns)**.
* Example output:

```python
(3, 2)
```

This means:

* **3 rows**
* **2 columns**

---

### `.ndim`

```python
print("Number of dimensions:", customers.ndim)
```

* Returns the number of dimensions (axes) in the array.
* A 2D array has:

```python
2
```

---

### `len()`

```python
print(len(customers))
```

* Returns the number of **rows** in the array.
* Example output:

```python
3
```

> **Note:** `len()` only counts the first dimension (rows). It does **not** return the total number of elements.

---

## Summary

| Function/Attribute | Purpose                          | Example Output         |
| ------------------ | -------------------------------- | ---------------------- |
| `print(customers)` | Displays the entire array        | `[['Alice' '25'] ...]` |
| `customers.shape`  | Returns `(rows, columns)`        | `(3, 2)`               |
| `customers.ndim`   | Returns the number of dimensions | `2`                    |
| `len(customers)`   | Returns the number of rows       | `3`                    |

### Quick Memory Tip

* **`.shape`** → *What is the size?* → `(rows, columns)`
* **`.ndim`** → *How many dimensions?* → `2`
* **`len()`** → *How many rows?* → First dimension only.

This note is a good foundation for learning 2D NumPy arrays before moving on to indexing, slicing, and vectorized operations.


````python
print("Customer Data:\n", customers)
print("Customers shape: ", customers.shape)
print("Number of dimensions: ", customers.ndim)
print("Total elements: ", customers.size)
print("Number of customers: ", len(customers))  # Number of rows
print("Number of features per customer: ", len(customers[0]))  # Number of columns
````

# Metrix multiplication

````python
A = np.array([
    [1,-2,-6],
    [3,4,-3],
    [3,1,2]
])

B = np.array([
    [1,4,3],
    [0,-2,3],
    [6,1,0]
])

print(f"A.B \n {A@B}")
print("================")
print(f"B.A \n {B@A}")
````

````python
# can use matmul function to calculate matrix multipication
print(f"A.B \n {np.matmul(A,B)}")
print("=================")
print(f"B.A \n {np.matmul(B,A)}")
````

````python

# .dot function can be used to calculations but not recommonded.

print(f"A.B \n {np.dot(A,B)}")
print("=================")
print(f"B.A \n {np.dot(B,A)}")
````
