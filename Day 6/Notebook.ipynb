```python
class ShoppingCart:
    def __init__(self, customer_name):
        self.customer_name = customer_name
        self.items = ["Laptop", "Mouse", "Keyboard"]

    # define the __len__ method to return the number of items in the cart (Magic Method | Dunder Method | Double Underscore Method )
    def __len__(self):
        return len(self.items)


my_cart = ShoppingCart("John Doe")
print(f"Customer: {my_cart.customer_name}")
print(f"Number of items in cart: {len(my_cart)}")

# print(my_cart[1])  # This will raise an error because __getitem__ is not defined
```

There fore , we can define the __getitem__ method to allow indexing into the ShoppingCart object. This will enable us to access items in the cart using square bracket notation.

```python
class ShoppingCart:
    def __init__(self, customer_name):
        self.customer_name = customer_name
        self.items = ["Laptop", "Mouse", "Keyboard"]

    def __len__(self):
        return len(self.items)

    # define the __getitem__ method to allow indexing into the items list
    def __getitem__(self, index):
        return self.items[index]


my_cart = ShoppingCart("John Doe")
print(f"Customer: {my_cart.customer_name}")
print(f"Number of items in cart: {len(my_cart)}")

print(f"Cart Item : {my_cart[1]} ")
```

 __call__ method is used to make an instance of a class callable like a function. It allows you to define the behavior of an object when it is called as if it were a function.

```python
#Example of __call__ method
class ShoppingCart:
    def __init__(self, customer_name):
        self.customer_name = customer_name
        self.items = ["Laptop", "Mouse", "Keyboard"]

    def __len__(self):
        return len(self.items)

    def __getitem__(self, index):
        return self.items[index]

    # define the __call__ method to allow the instance to be called like a function
    def __call__(self):
        return f"Shopping Cart for {self.customer_name} contains: {', '.join(self.items)}"
    
my_cart = ShoppingCart("John Doe")
print(f"Customer: {my_cart.customer_name}")
print(f"Number of items in cart: {len(my_cart)}")
print(f"Cart Item : {my_cart[1]} ")

# Calling the instance like a function
print(my_cart())  # This will invoke the __call__ method and print the cart details
```

__repr__ method is used to define the string representation of an object.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
person = Person("Alice", 30)
# Without __repr__, printing the object will show a default representation
print(person)  # Output: <__main__.Person object at 0x...>
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def __repr__(self):
        return f"Person(name='{self.name}', age={self.age})"
    
person = Person("Alice", 30)

print(person)
```


__str__ function is used to get a user-friendly string representation of an object. It is often used for displaying information to the end user. The __str__ method is called when you use the str() function or print the object.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # this method is called when you use the str() function or print the object
    # in top level of hierarchy, it is used to provide a user-friendly string representation of the object
    def __str__(self):
        return f"From Str   : {self.name}, {self.age} years old"
    
    def __repr__(self):
        return f"From Repre :Person(name='{self.name}', age={self.age})"
    
person = Person("Alice", 30)
print(person)  # Output: From Str   : Alice, 30 years old
```
