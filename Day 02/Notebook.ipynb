# Logical operators
Logical operators are used to combine conditional statements. The three main logical operators are:
- and: Returns True if both statements are true
- or: Returns True if at least one statement is true
- not: Reverse the result, returns False if the result is true


````python
# Examples for logical operators
x = 5
y = 10
print(x > 3 and y < 15) # True
print(x > 3 or y < 5)   # True
print(not (x > 3 and y < 15)) # False
````

# Conditional Statements

Conditional statements are decision-making structures in Python that allow a program to execute different blocks of code based on whether a specified condition is true or false. They help control the flow of a program by enabling it to respond to different inputs, values, or situations.

Unlike many programming languages that use curly braces `{}` to define code blocks, **Python uses indentation (spaces or tabs)** to indicate which statements belong to a particular block. All statements with the same indentation level are considered part of the same block of code.

Python provides three main conditional statements:

* **`if`** – Executes a block of code when a condition is true.
* **`if-else`** – Executes one block when the condition is true and another block when it is false.
* **`if-elif-else`** – Checks multiple conditions and executes the corresponding block for the first true condition.

### Example

```python
age = 18

if age >= 18:
    print("Adult")      # Indented block
    print("Can vote")   # Same block

print("Program ended")  # Outside the block
```

In the example above, the two indented `print()` statements belong to the `if` block. The last `print()` statement is not indented, so it is outside the conditional block and executes regardless of the condition.

Conditional statements are essential for creating dynamic and interactive programs, as they allow applications to make decisions and perform actions based on changing conditions.


````python
# Examples for conditional statements
age = 20
if age < 18:
    print("You are a minor.")
elif age < 65:
    print("You are an adult.")
else:
    print("You are a senior.")
````

# Data Structures Overview

Python provides **4 main collection data types** for storing multiple values:

1. **List**
2. **Tuple**
3. **Set**
4. **Dictionary (Dict)**

Note: 
An ordered data structure stores elements in a specific sequence, so each element has a position (index). Because positions are defined, you can access elements using their index.

## Creating Collections

### List

```python
numbers = [1, 2, 3, 4]
```

**Special Points:**

* Ordered collection.
* Allows duplicate values.
* Mutable (can be modified after creation).

### Tuple

```python
numbers = (1, 2, 3, 4)
```

**Special Points:**

* Ordered collection.
* Allows duplicate values.
* Immutable (cannot be modified after creation).

### Set

```python
numbers = {1, 2, 3, 4}
```

**Special Points:**

* Unordered collection.
* Does not allow duplicate values.
* Mutable (can add or remove items).

### Dictionary

```python
student = {
    "name": "John",
    "age": 20
}
```

**Special Points:**

* Stores data as **key-value pairs**.
* Keys must be unique.
* Mutable and ordered (Python 3.7+).

## Comparison Table

| Data Structure | Ordered | Duplicates | Mutable |
| -------------- | ------- | ---------- | ------- |
| List           | Yes     | Yes        | Yes     |
| Tuple          | Yes     | Yes        | No      |
| Set            | No      | No         | Yes     |
| Dictionary     | Yes     | Keys: No   | Yes     |

## Summary

* Use **List** when data needs to be changed frequently.
* Use **Tuple** for fixed data that should not change.
* Use **Set** when unique values are required.
* Use **Dictionary** when data should be stored as key-value pairs.


````python

example_list = [1, 2, 3, 4, 5]
example_tuple = (1, 2, 3, 4, 5)
example_set = {1, 2, 3, 4, 5}
example_dict = {'a': 1, 'b': 2, 'c': 3}

print(type(example_list))  # <class 'list'>
print(type(example_tuple)) # <class 'tuple'>    
print(type(example_set))   # <class 'set'>
print(type(example_dict))  # <class 'dict'>
````

````python

# ----------------------------
# CREATING COLLECTIONS
# ----------------------------
# List → mutable (can be changed)
my_list = [1, 2, 3]

# Tuple → immutable (cannot be changed)
my_tuple = (1, 2, 3)

print("Initial List:", my_list)
print("Initial Tuple:", my_tuple)


# ----------------------------
# ACCESSING ELEMENTS
# ----------------------------
# Both list and tuple support index-based access
print("List[0]:", my_list[0])     # 1
print("Tuple[0]:", my_tuple[0])   # 1


# ----------------------------
# UPDATING VALUES
# ----------------------------
# List allows modification
my_list[0] = 100
print("After updating list[0] = 100:", my_list)   # [100, 2, 3]

# Tuple does NOT allow modification
# my_tuple[0] = 10  ❌ TypeError


# ----------------------------
# REMOVING / MODIFYING VALUES (LIST ONLY)
# ----------------------------

# Reset value
my_list[0] = 1
print("After resetting list:", my_list)   # [1, 2, 3]

# Remove element by index
my_list.pop(0)
print("After pop(0):", my_list)   # [2, 3]

# Remove element by value
my_list.remove(2)
print("After remove(2):", my_list)   # [3]


# ----------------------------
# ADDING VALUES (LIST ONLY)
# ----------------------------

# Add value at end
my_list.append(100)
print("After append(100):", my_list)   # [3, 100]

# Add value at specific position
my_list.insert(1, 50)
print("After insert(1, 50):", my_list)   # [3, 50, 100]



# Tuple cannot be modified
# my_tuple.append(10) ❌ AttributeError
````

````python

example_list = [1, 2, 3]
print(example_list)
print (f"Id of the list {id(example_list)}")

example_list.append(4)
print(example_list)
print (f"Id of the list {id(example_list)}")
# Memory address of the list remains same after modification, because list is mutable.

example_tuple = (1, 2, 3)
print(example_tuple)
print (f"Id of the tuple {id(example_tuple)}")

example_tuple = example_tuple + (4,)
print(example_tuple)
print (f"Id of the tuple {id(example_tuple)}")

# Memory address of the tuple changes after modification, because tuple is immutable.
````

# Intersection of 2 Sets

The intersection of two sets contains only the elements that are present in BOTH sets.
In Python, we use the .intersection() method or the & operator to find common elements.

Example:
```python
 set_A = {10, 12, 5, 3, 8}
 set_B = {8, 15, 20, 10}
 Intersection = {10, 8}  (only these elements exist in both sets)
```
Methods to find intersection:
- 1. Using .intersection() method: set_A.intersection(set_B)
- 2. Using & operator: set_A & set_B

````python
set_A = {10,12,5,3,8}
set_B = {8,15,20,10}

common_elements = set_A.intersection(set_B)
print(f"Common elements between set_A and set_B: {common_elements}")
print(f"Common elements between set_A and set_B: {set_A & set_B}")
````

# Union of 2 Sets

The union of two sets contains **all unique elements** from both sets (duplicates are removed automatically).

In Python, we use the `.union()` method or the `|` operator to find the union of sets.

Example:

```python
set_A = {10, 12, 5, 3, 8}
set_B = {8, 15, 20, 10}

Union = {3, 5, 8, 10, 12, 15, 20}
# all unique elements from both sets
```

Methods to find union:

* 1. Using `.union()` method: `set_A.union(set_B)`
* 2. Using `|` operator: `set_A | set_B`


````python
set_A = {10,12,5,3,8}
set_B = {8,15,20,10}

union_elements = set_A.union(set_B)
print(f"Union of set_A and set_B: {union_elements}")
print(f"Union of set_A and set_B: {set_A | set_B}")
````
